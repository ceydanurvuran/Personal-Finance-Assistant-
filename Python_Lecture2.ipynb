## 1. Python Lists

Lists are used to store multiple items in a single variable.  
Lists are one of 4 built-in data types in Python used to store collections of data. The other 3 are Tuple, Set, and Dictionary.

A list is created using square brackets `[]`:

```python
thislist = ["apple", "banana", "cherry"]
print(thislist)
```

### List Properties

- **Ordered:** Items have a defined order, and that order will not change.
- **Changeable (Mutable):** We can change, add, and remove items after the list has been created.
- **Allow Duplicates:** Lists can have items with the same value.
- **Indexed:** The first item has index `[0]`, the second item has index `[1]`, etc.

```python
thislist = ["apple", "banana", "cherry", "apple", "cherry"]
print(thislist)       # duplicates are allowed
print(len(thislist))  # 5
```

### 1.1 Access List Items

List items are indexed. You can access them by referring to the index number:

```python
thislist = ["apple", "banana", "cherry"]
print(thislist[1])    # banana
print(thislist[-1])   # cherry (negative indexing: start from the end)
```

**Range of Indexes (Slicing):**

```python
thislist = ["apple", "banana", "cherry", "orange", "kiwi", "melon", "mango"]
print(thislist[2:5])   # ['cherry', 'orange', 'kiwi']
print(thislist[:4])    # ['apple', 'banana', 'cherry', 'orange']
print(thislist[2:])    # ['cherry', 'orange', 'kiwi', 'melon', 'mango']
```

**Check if Item Exists:**

```python
if "apple" in thislist:
    print("Yes, 'apple' is in the list")
```

### 1.3 Add List Items

**`append()`** - Adds an item to the end of the list:

```python
thislist = ["apple", "banana", "cherry"]
thislist.append("orange")
print(thislist)  # ['apple', 'banana', 'cherry', 'orange']
```

**`insert()`** - Adds an item at a specified index:

```python
thislist = ["apple", "banana", "cherry"]
thislist.insert(1, "orange")
print(thislist)  # ['apple', 'orange', 'banana', 'cherry']
```

**`extend()`** - Adds elements from another list (or any iterable) to the current list:

```python
thislist = ["apple", "banana", "cherry"]
tropical = ["mango", "pineapple", "papaya"]
thislist.extend(tropical)
print(thislist)  # ['apple', 'banana', 'cherry', 'mango', 'pineapple', 'papaya']
```

### 1.4 Remove List Items

**`remove()`** - Removes the first occurrence of the specified item:

```python
thislist = ["apple", "banana", "cherry"]
thislist.remove("banana")
print(thislist)  # ['apple', 'cherry']
```

**`pop()`** - Removes the item at the specified index (or the last item if index is not specified):

```python
thislist = ["apple", "banana", "cherry"]
thislist.pop(1)
print(thislist)  # ['apple', 'cherry']

thislist.pop()   # removes last item
print(thislist)  # ['apple']
```

**`del`** - Removes the item at the specified index, or deletes the entire list:

```python
thislist = ["apple", "banana", "cherry"]
del thislist[0]
print(thislist)  # ['banana', 'cherry']
```

**`clear()`** - Empties the list (the list still exists but has no content):

```python
thislist = ["apple", "banana", "cherry"]
thislist.clear()
print(thislist)  # []
```

### 1.6 List Comprehension

List comprehension offers a shorter syntax to create a new list based on the values of an existing list.

**Syntax:** `newlist = [expression for item in iterable if condition]`

```python
fruits = ["apple", "banana", "cherry", "kiwi", "mango"]

# Without list comprehension
newlist = []
for x in fruits:
    if "a" in x:
        newlist.append(x)
print(newlist)  # ['apple', 'banana', 'mango']

# With list comprehension
newlist = [x for x in fruits if "a" in x]
print(newlist)  # ['apple', 'banana', 'mango']
```

**More Examples:**

```python
# Create a list of squares
squares = [x**2 for x in range(10)]
print(squares)  # [0, 1, 4, 9, 16, 25, 36, 49, 64, 81]

# Convert to uppercase
upper_fruits = [x.upper() for x in fruits]
print(upper_fruits)
```

### 1.7 Sort Lists

The `sort()` method sorts the list alphanumerically, ascending by default:

```python
thislist = ["orange", "mango", "kiwi", "pineapple", "banana"]
thislist.sort()
print(thislist)  # ['banana', 'kiwi', 'mango', 'orange', 'pineapple']

# Sort descending
thislist.sort(reverse=True)
print(thislist)  # ['pineapple', 'orange', 'mango', 'kiwi', 'banana']

# Numeric sort
numbers = [100, 50, 65, 82, 23]
numbers.sort()
print(numbers)   # [23, 50, 65, 82, 100]
```

**`sorted()`** returns a new sorted list without changing the original:

```python
original = [5, 3, 8, 1]
new_list = sorted(original)
print(original)  # [5, 3, 8, 1] (unchanged)
print(new_list)  # [1, 3, 5, 8]
```

### 1.8 Copy & Join Lists

You **cannot** copy a list simply by typing `list2 = list1`, because `list2` will only be a **reference** to `list1`.  
Changes made in `list1` will automatically also be made in `list2`.

**Use `copy()` method:**

```python

mylist = thislist.copy()

```

**Join Two Lists:**

```python
list1 = ["a", "b", "c"]
list2 = [1, 2, 3]
list3 = list1 + list2
print(list3)  # ['a', 'b', 'c', 1, 2, 3]
```

## 2. Python Tuples

Tuples are used to store multiple items in a single variable.  
A tuple is a collection which is **ordered** and **unchangeable** (immutable).

Tuples are written with round brackets `()`:

```python
thistuple = ("apple", "banana", "cherry")
print(thistuple)
```

### Tuple Properties

- **Ordered:** Items have a defined order, and that order will not change.
- **Unchangeable (Immutable):** We cannot change, add, or remove items after the tuple has been created.
- **Allow Duplicates:** Tuples can have items with the same value.
- **Indexed:** The first item has index `[0]`.

```python
thistuple = ("apple", "banana", "cherry", "apple", "cherry")
print(thistuple)
print(len(thistuple))  # 5
```

> **Important:** To create a tuple with only one item, you need to add a comma after the item:

```python
thistuple = ("apple",)   # This is a tuple
print(type(thistuple))   # <class 'tuple'>

not_a_tuple = ("apple")  # This is NOT a tuple, it's a string
print(type(not_a_tuple)) # <class 'str'>
```

### 2.1 Access Tuple Items

You can access tuple items by referring to the index number (same as lists):

```python
thistuple = ("apple", "banana", "cherry")
print(thistuple[1])    # banana
print(thistuple[-1])   # cherry
print(thistuple[0:2])  # ('apple', 'banana')
```

### 2.2 Update Tuples (Workaround)

Tuples are unchangeable, but there is a workaround: convert the tuple into a list, make changes, and convert it back.

```python
x = ("apple", "banana", "cherry")
y = list(x)          # convert to list
y[1] = "kiwi"        # change the item
x = tuple(y)         # convert back to tuple
print(x)             # ('apple', 'kiwi', 'cherry')
```

**Adding items (workaround):**

```python
thistuple = ("apple", "banana", "cherry")
y = list(thistuple)
y.append("orange")
thistuple = tuple(y)
print(thistuple)  # ('apple', 'banana', 'cherry', 'orange')
```

**Or use `+=` to add tuples together:**

```python
thistuple = ("apple", "banana", "cherry")
thistuple += ("orange",)
print(thistuple)  # ('apple', 'banana', 'cherry', 'orange')
```

### 2.3 Unpack Tuples

When we create a tuple, we normally assign values to it — this is called **packing**.  
In Python, we are also allowed to extract the values back into variables — this is called **unpacking**.

```python
fruits = ("apple", "banana", "cherry")
(green, yellow, red) = fruits

print(green)   # apple
print(yellow)  # banana
print(red)     # cherry
```

> **Note:** The number of variables must match the number of values in the tuple.

**Using Asterisk `*`:**

If the number of variables is less than the number of values, you can add an `*` to collect the remaining values as a list:

```python
fruits = ("apple", "banana", "cherry", "strawberry", "raspberry")
(green, yellow, *red) = fruits

print(green)   # apple
print(yellow)  # banana
print(red)     # ['cherry', 'strawberry', 'raspberry']
```

### 2.4 Loop Tuples & Tuple Methods

**Loop through a tuple:**

```python
thistuple = ("apple", "banana", "cherry")
for x in thistuple:
    print(x)
```

**Join Tuples:**

```python
tuple1 = ("a", "b", "c")
tuple2 = (1, 2, 3)
tuple3 = tuple1 + tuple2
print(tuple3)  # ('a', 'b', 'c', 1, 2, 3)

# Multiply
mytuple = ("apple", "banana", "cherry")
doubled = mytuple * 2
print(doubled)  # ('apple', 'banana', 'cherry', 'apple', 'banana', 'cherry')
```

**Tuple Methods:**

| Method | Description |
|---|---|
| `count()` | Returns the number of times a specified value occurs |
| `index()` | Searches for a specified value and returns the index |

```python
thistuple = (1, 3, 7, 8, 7, 5, 4, 6, 8, 5)
print(thistuple.count(5))  # 2
print(thistuple.index(8))  # 3 (first occurrence)
```

## 3. Python Sets

Sets are used to store multiple items in a single variable.  
A set is a collection which is **unordered**, **unchangeable***, and **unindexed**.

> *Set items are unchangeable, but you can add and remove items.

Sets are written with curly brackets `{}`:

```python
thisset = {"apple", "banana", "cherry"}
print(thisset)
```

### Set Properties

- **Unordered:** Items have no defined order. They can appear in a different order every time.
- **No Duplicates:** Sets cannot have two items with the same value.
- **Unindexed:** You cannot access items by index or key.

```python
thisset = {"apple", "banana", "cherry", "apple"}  # duplicate 'apple' will be ignored
print(thisset)       # {'banana', 'cherry', 'apple'}
print(len(thisset))  # 3
```

> **Note:** The values `True` and `1` are considered the same value in sets (also `False` and `0`).

### 3.1 Access Set Items

You **cannot** access items in a set by referring to an index, because sets are unordered.  
But you can loop through the set or check if an item exists using the `in` keyword:

```python
thisset = {"apple", "banana", "cherry"}

for x in thisset:
    print(x)

print("banana" in thisset)  # True
```

### 3.2 Add Set Items

**`add()`** - Adds one item to the set:

```python
thisset = {"apple", "banana", "cherry"}
thisset.add("orange")
print(thisset)
```

**`update()`** - Adds items from another set (or any iterable) into the current set:

```python
thisset = {"apple", "banana", "cherry"}
tropical = {"pineapple", "mango", "papaya"}
thisset.update(tropical)
print(thisset)

# You can also add a list to a set
thisset = {"apple", "banana", "cherry"}
mylist = ["kiwi", "orange"]
thisset.update(mylist)
print(thisset)
```

### 3.3 Remove Set Items

**`remove()`** - Removes the specified item. Raises an error if the item does not exist:

```python
thisset = {"apple", "banana", "cherry"}
thisset.remove("banana")
print(thisset)
```

**`discard()`** - Removes the specified item. Does NOT raise an error if the item does not exist:

```python
thisset = {"apple", "banana", "cherry"}
thisset.discard("banana")
print(thisset)
```

**`pop()`** - Removes a random item (since sets are unordered):

```python
thisset = {"apple", "banana", "cherry"}
x = thisset.pop()
print(x)        # removed item
print(thisset)  # remaining set
```

**`clear()`** - Empties the set:

```python
thisset = {"apple", "banana", "cherry"}
thisset.clear()
print(thisset)  # set()
```

### 3.4 Join Sets

**`union()`** - Returns a new set containing all items from both sets (no duplicates):

```python
set1 = {"a", "b", "c"}
set2 = {1, 2, 3}
set3 = set1.union(set2)
print(set3)  # {1, 2, 3, 'a', 'c', 'b'}

# Or use the | operator
set3 = set1 | set2
```

**`intersection()`** - Returns a set containing only the items that are in both sets:

```python
set1 = {"apple", "banana", "cherry"}
set2 = {"google", "microsoft", "apple"}
set3 = set1.intersection(set2)
print(set3)  # {'apple'}

# Or use the & operator
set3 = set1 & set2
```

**`difference()`** - Returns a set containing items that are in the first set but not in the second:

```python
set1 = {"apple", "banana", "cherry"}
set2 = {"google", "microsoft", "apple"}
set3 = set1.difference(set2)
print(set3)  # {'banana', 'cherry'}

# Or use the - operator
set3 = set1 - set2
```

**`symmetric_difference()`** - Returns a set containing items that are in either set, but not in both:

```python
set1 = {"apple", "banana", "cherry"}
set2 = {"google", "microsoft", "apple"}
set3 = set1.symmetric_difference(set2)
print(set3)  # {'banana', 'cherry', 'google', 'microsoft'}

# Or use the ^ operator
set3 = set1 ^ set2
```

## 4. Python Dictionaries

Dictionaries are used to store data values in **key:value** pairs.  
A dictionary is a collection which is **ordered** (as of Python 3.7), **changeable**, and **does not allow duplicate keys**.

Dictionaries are written with curly brackets `{}`, with key:value pairs:

```python
thisdict = {
    "brand": "Ford",
    "model": "Mustang",
    "year": 1964
}
print(thisdict)
print(len(thisdict))  # 3
```

### Dictionary Properties

- **Ordered:** Items have a defined order (since Python 3.7).
- **Changeable (Mutable):** We can change, add, and remove items.
- **No Duplicate Keys:** Two items cannot have the same key.
- **Keys can be** any immutable type (strings, numbers, tuples).
- **Values can be** any data type.

### 4.1 Access Dictionary Items

You can access the items by referring to the key name:

```python
thisdict = {"brand": "Ford", "model": "Mustang", "year": 1964}

x = thisdict["model"]
print(x)  # Mustang

# Or use the get() method
x = thisdict.get("model")
print(x)  # Mustang
```

**Get All Keys:**

```python
print(thisdict.keys())    # dict_keys(['brand', 'model', 'year'])
```

**Get All Values:**

```python
print(thisdict.values())  # dict_values(['Ford', 'Mustang', 1964])
```

**Get All Key:Value Pairs:**

```python
print(thisdict.items())   # dict_items([('brand', 'Ford'), ('model', 'Mustang'), ('year', 1964)])
```

**Check if Key Exists:**

```python
if "model" in thisdict:
    print("Yes, 'model' is one of the keys")
```

### 4.2 Change & Add Dictionary Items

**Change an item** by referring to its key:

```python
thisdict = {"brand": "Ford", "model": "Mustang", "year": 1964}
thisdict["year"] = 2018
print(thisdict)  # {'brand': 'Ford', 'model': 'Mustang', 'year': 2018}
```

**`update()`** - Updates the dictionary with the given key:value pairs:

```python
thisdict.update({"year": 2020})
```

**Add a new item** by using a new key:

```python
thisdict = {"brand": "Ford", "model": "Mustang", "year": 1964}
thisdict["color"] = "red"
print(thisdict)  # {'brand': 'Ford', 'model': 'Mustang', 'year': 1964, 'color': 'red'}
```

### 4.3 Remove Dictionary Items

**`pop()`** - Removes the item with the specified key:

```python
thisdict = {"brand": "Ford", "model": "Mustang", "year": 1964}
thisdict.pop("model")
print(thisdict)  # {'brand': 'Ford', 'year': 1964}
```

**`popitem()`** - Removes the last inserted item:

```python
thisdict = {"brand": "Ford", "model": "Mustang", "year": 1964}
thisdict.popitem()
print(thisdict)  # {'brand': 'Ford', 'model': 'Mustang'}
```

**`del`** - Removes the item with the specified key:

```python
thisdict = {"brand": "Ford", "model": "Mustang", "year": 1964}
del thisdict["model"]
print(thisdict)  # {'brand': 'Ford', 'year': 1964}
```

**`clear()`** - Empties the dictionary:

```python
thisdict.clear()
print(thisdict)  # {}
```

### 4.4 Loop Through a Dictionary

**Loop through keys:**

```python
thisdict = {"brand": "Ford", "model": "Mustang", "year": 1964}

for key in thisdict:
    print(key)
```

**Loop through values:**

```python
for value in thisdict.values():
    print(value)
```

**Loop through key:value pairs:**

```python
for key, value in thisdict.items():
    print(key, ":", value)
```

### 4.5 Copy & Nested Dictionaries

**Copy a dictionary** (same issue as lists — use `copy()` or `dict()`):

```python
thisdict = {"brand": "Ford", "model": "Mustang", "year": 1964}
mydict = thisdict.copy()
# or
mydict = dict(thisdict)
```

**Nested Dictionaries:**

A dictionary can contain dictionaries — this is called nested dictionaries:

```python
myfamily = {
    "child1": {
        "name": "Emil",
        "year": 2004
    },
    "child2": {
        "name": "Tobias",
        "year": 2007
    },
    "child3": {
        "name": "Linus",
        "year": 2011
    }
}

# Access nested item
print(myfamily["child2"]["name"])  # Tobias
```

### 4.6 Dictionary Methods Summary

| Method | Description |
|---|---|
| `clear()` | Removes all the elements from the dictionary |
| `copy()` | Returns a copy of the dictionary |
| `fromkeys()` | Returns a dictionary with the specified keys and value |
| `get()` | Returns the value of the specified key |
| `items()` | Returns a list of tuples containing key:value pairs |
| `keys()` | Returns a list of all the keys |
| `pop()` | Removes the element with the specified key |
| `popitem()` | Removes the last inserted key:value pair |
| `setdefault()` | Returns the value of the specified key. If it does not exist, inserts the key with a value |
| `update()` | Updates the dictionary with the specified key:value pairs |
| `values()` | Returns a list of all the values |

### Question 3: Common & Unique Skills Finder

A company has 3 teams. Each team has a **tuple** containing the team name and a **set** of skills that team members have:

```python
teams = [
    ("Frontend", {"HTML", "CSS", "JavaScript", "React", "TypeScript"}),
    ("Backend", {"Python", "Java", "SQL", "Docker", "JavaScript"}),
    ("Data Science", {"Python", "SQL", "R", "TensorFlow", "Statistics"})
]
```

Write a program that:

1. Finds the **set of all skills** across all teams (union of all skill sets).
2. Finds the skills that are **common to all 3 teams** (intersection of all skill sets).
3. For **each team**, finds the skills that are **unique to that team** (skills no other team has).
4. Stores the results in a **dictionary** where the key is the team name and the value is the set of unique skills.
5. Prints everything.

**Expected Output (approximately):**
```
All Skills: {'HTML', 'CSS', 'JavaScript', 'React', 'TypeScript', 'Python', 'Java', 'SQL', 'Docker', 'R', 'TensorFlow', 'Statistics'}
Common Skills: set()  (no skill is shared by ALL 3 teams)
Unique Skills:
  Frontend: {'HTML', 'CSS', 'React', 'TypeScript'}
  Backend: {'Java', 'Docker'}
  Data Science: {'R', 'TensorFlow', 'Statistics'}
```

**Hint:** Use set operations like `union()`, `intersection()`, and `difference()`.

````python
# Write your solution here

teams = [
    ("Frontend", {"HTML", "CSS", "JavaScript", "React", "TypeScript"}),
    ("Backend", {"Python", "Java", "SQL", "Docker", "JavaScript"}),
    ("Data Science", {"Python", "SQL", "R", "TensorFlow", "Statistics"})
]

all_skills=set()

for team,skills in teams:
    all_skills=all_skills.union(skills)

print(all_skills)

common_skills = teams[0][1]

for team, skills in teams[1:]:
    common_skills = common_skills.intersection(skills)

print(common_skills)

unique = {}

for i, (team, skills) in enumerate(teams):
    other_skills = set()

    for j, (t, s) in enumerate(teams):
        if i != j:
            other_skills = other_skills.union(s)

    unique[team] = skills.difference(other_skills)

print(unique)
````
